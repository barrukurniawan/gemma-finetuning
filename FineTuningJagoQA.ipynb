```python
###### UBAH DATASET KE FORMAT CHAT & UBAH KE DATASET BARU FORMAT HUGGINGFACE #####
from datasets import Dataset

def to_chat_format(item): # fungsi untuk format input jadi gaya chat model
    return {
        "text": (
            f"<start_of_turn>user\n{item['prompt']}<end_of_turn>" # Token yang menunjukkan awal giliran dialog:
            f"<start_of_turn>model\n{item['completion']}<end_of_turn>" # Token yang menunjukkan akhir giliran dialog:
        )
    }

data_raw = json.load(open("qa_data.json"))
data_formatted = [to_chat_format(item) for item in data_raw] # mapping setiap item ke format chat

dataset = Dataset.from_list(data_formatted)
dataset.save_to_disk("qa_dataset_question") # simpan dataset HuggingFace dari list
print("✔ Dataset terformat disimpan!")
```

```python
##### LOAD MODEL GEMMA 1B DARI UNSLOTH #####
''' Jika ingin mengganti model bisa cek https://huggingface.co/unsloth/collections alasan menggunakan gemma 1B
karena mencari performa yang ringan dengan 1B yang berarti 1 miliar parameter '''

from unsloth import FastModel
import torch

model_name = "unsloth/gemma-3-1b-it"

model, tokenizer = FastModel.from_pretrained(
    model_name=model_name,
    load_in_4bit=True, # rekomendasi quantization 4bit di True supaya hemat VRAM
    full_finetuning=False, # hanya LoRA (Low-Rank Adaptation) teknik efisiensi memori
    max_seq_length=2048
)
```

```python
##### KONFIGURASI LORA #####
''' Baca https://medium.com/@fartypantsham untuk mengetahui tentang Lora lebih lengkap dan ringkas
efisien untuk bantu training model dengan parameter yang kecil '''
model = FastModel.get_peft_model(
    model,
    finetune_language_layers=True,
    finetune_attention_modules=True,
    finetune_mlp_modules=True,
    r=32, # rank lora, kita naikkan dari yang sebelumnya 16
    lora_alpha=32, # scaling LoRA
    target_modules=[ # modul target yang dioptimalkan LoRA
        "q_proj", "k_proj", "v_proj",
        "o_proj", "gate_proj",
        "up_proj", "down_proj"
    ],
    lora_dropout=0,
    use_gradient_checkpointing="unsloth", # hemat VRAM
    bias="none",
    random_state=3407
)
```

```python
##### SETUP TRAINER #####
''' kalian akan sering dengar soal tokenizer pada LLM. Tokenizer : mengubah Teks → Token (angka),
hasilnya juga dibalikkan dari Token → Teks'''
from trl import SFTTrainer, SFTConfig
from datasets import Dataset

trainer = SFTTrainer(
    model=model, # model Gemma + LoRA, merujuk ke load model gemma yang sebelumnya
    tokenizer=tokenizer,
    train_dataset=dataset,
    args=SFTConfig(
        dataset_text_field="text",
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        num_train_epochs=3, # training selama 3 epoch
        max_steps=500, # total step training, karena dataset kita ada 160, jadi cukup 500 maksimal
        warmup_steps=10,
        learning_rate=1e-4,
        logging_steps=1,
        optim="adamw_8bit",
        weight_decay=0.001,
        lr_scheduler_type="linear",
        seed=3407,
        report_to="none",
    ),
)
```

```python
##### MEMULAI TRAINING MODEL DENGAN DATASET BARU #####
print("✔ Training dimulai...")
trainer.train()
print("✔ Training selesai!")
```
